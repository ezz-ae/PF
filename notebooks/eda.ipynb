# Marketplace Integrity Signals
This notebook quickly inspects the sample listing data included with the engine.

```python
from pathlib import Path
import json

data_path = Path(__file__).resolve().parents[1] / 'data' / 'sample_listings.json'
listings = json.loads(data_path.read_text())

for listing in listings:
    details = listing.get('details', {})
    community = details.get('community', 'unknown')
    print(f"{listing['listing_id']} · {community} · {len(listing.get('images', []))} images")
```
